```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams['figure.dpi'] = 300
fontsize = 8
linewidth = 0.5
```

```python
def gaussian(
    support, # domain of the function
    ctr, # mean of the function
    amp, # amplitude of the function
    mid, # midline of the function
    std # bandwidth parameter
):

    f = np.exp(-1*(((support-ctr)**2)/(2*std**2)))

    return mid+(amp*f)
```

## Calculating response onset times

```python
T = np.arange(0,250)
exc = gaussian(T, 85, 1, 0, 20)
inh = gaussian(T, 135, 1, 0, 20)
noise = np.random.normal(0,0.05,250)
```

```python

fig, ax = plt.subplots(figsize = (4,2))
ax.tick_params(labelsize = fontsize)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(exc-inh+noise, color = 'black', linewidth = 0.75)
ax.set_ylabel("Norm. firing rate", fontsize = fontsize)
ax.set_xlabel("Time (ms)", fontsize = fontsize)
ax.set_title("Response to a single SF condition", fontsize = fontsize)
```

```python

fig, ax = plt.subplots(figsize = (4,2))
ax.tick_params(labelsize = fontsize)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(exc-inh+noise, color = 'black', linewidth = 0.75)
ax.plot([25]*250, np.linspace(-1,1,250), '--', linewidth = linewidth, color = 'black')
ax.set_ylabel("Norm. firing rate", fontsize = fontsize)
ax.set_xlabel("Time (ms)", fontsize = fontsize)
ax.set_title("Response to a single SF condition", fontsize = fontsize)
```

```python

fig, ax = plt.subplots(figsize = (4,2))
ax.tick_params(labelsize = fontsize)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(exc-inh+noise, color = 'black', linewidth = 0.75)

r = exc-inh+noise
inc_thresh = 6*(r[:25].std())

ax.plot(np.full(250, inc_thresh), '--', linewidth = linewidth, color = 'red', label = 'inclusion thresh')
ax.plot(np.full(250, -inc_thresh), '--', linewidth = linewidth, color = 'red')
ax.legend(fontsize = 6, frameon = False)

ax.set_ylabel("Norm. firing rate", fontsize = fontsize)
ax.set_xlabel("Time (ms)", fontsize = fontsize)
ax.set_title("Response to a single SF condition", fontsize = fontsize)
```

```python

fig, ax = plt.subplots(figsize = (4,2))
ax.tick_params(labelsize = fontsize)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(exc-inh+noise, color = 'black', linewidth = 0.75)

r = exc-inh+noise
inc_thresh = 6*(r[:25].std())

ax.plot(np.full(250, inc_thresh), '--', linewidth = linewidth, color = 'red', label = 'inclusion thresh')
ax.plot(np.full(250, -inc_thresh), '--', linewidth = linewidth, color = 'red')

ax.plot(np.full(250, inc_thresh/3), '--', linewidth = linewidth, color = 'green', label = 'onset thresh')
ax.plot(np.full(250, -inc_thresh/3), '--', linewidth = linewidth, color = 'green')

ax.legend(fontsize = 6, frameon = False)

ax.set_ylabel("Norm. firing rate", fontsize = fontsize)
ax.set_xlabel("Time (ms)", fontsize = fontsize)
ax.set_title("Response to a single SF condition", fontsize = fontsize)
```
