```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.rcParams['figure.dpi'] = 300
fontsize = 8
linewidth = 0.5
```

```python
def gaussian(
    support, # domain of the function
    ctr, # mean of the function
    amp, # amplitude of the function
    mid, # midline of the function
    std # bandwidth parameter
):

    f = np.exp(-1*(((support-ctr)**2)/(2*std**2)))

    return mid+(amp*f)
```

## Calculating response onset times

```python
T = np.arange(0,250)
exc = gaussian(T, 85, 1, 0, 20)
inh = gaussian(T, 135, 1, 0, 20)
noise = np.random.normal(0,0.05,250)
```

```python

fig, ax = plt.subplots(figsize = (4,2))
ax.tick_params(labelsize = fontsize)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(exc-inh+noise, color = 'black', linewidth = 0.75)
ax.set_ylabel("Norm. firing rate", fontsize = fontsize)
ax.set_xlabel("Time (ms)", fontsize = fontsize)
ax.set_title("Response to a single SF condition", fontsize = fontsize)
```

```python
def sort_by_sf_tuning(M, sortby = 'max'):
    """
    Sorts a matrix of (N(neurons) x N(SF)) by
    lowest to highest SF tuning. 
    """
        
    M_dict = {i:[] for i in range(10)}
    for row in M:
        sort_mode = {
            'max': row.max(),
            'min': row.min()
        }
        i = np.where(row == sort_mode[sortby])[0][0]
        M_dict[i].append(row)
    for key in M_dict.keys():
        M_dict[key] = np.array(M_dict[key])
    M = [M_dict[key] for key in M_dict.keys()]
    return np.concatenate(M, axis = 0)
```

```python
sfs = [0.02,0.04,0.08,0.1,0.12,0.16,0.2,0.24,0.28,0.32]
```
